```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Item Visibility'], df['Sales'], alpha=0.5, color='purple', edgecolor='none')
plt.title('Item Visibility vs. Sales')
plt.xlabel('Item Visibility')
plt.ylabel('Sales')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# Group data by Fat Content and calculate mean sales
fat_sales = df.groupby('Item Fat Content')['Sales'].mean()

plt.figure(figsize=(6, 5))
plt.bar(fat_sales.index, fat_sales.values, color='salmon', edgecolor='black', width=0.5)
plt.title('Average Sales by Item Fat Content')
plt.xlabel('Item Fat Content')
plt.ylabel('Average Sales')
plt.show()
```

```python
# Group data by Year and sum up the sales
yearly_sales = df.groupby('Outlet Establishment Year')['Sales'].sum()

plt.figure(figsize=(10, 5))
plt.plot(yearly_sales.index, yearly_sales.values, marker='o', color='darkblue', linewidth=2)
plt.title('Total Sales by Outlet Establishment Year')
plt.xlabel('Establishment Year')
plt.ylabel('Total Sales')
plt.grid(True, alpha=0.3)
plt.show()
```
